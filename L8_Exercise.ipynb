```r
# Read your data here; 

# Clean it up, reformat it, filter it, sort it, group it, etc.

# Create your visualizations and save them as png files, then prepare your final pdf document elsewhere

# <- ALL YOUR CODE BELOW THIS POINT ->
Gen <- read.csv("./generation.csv",as.is=TRUE, header=TRUE)
population<- read.csv("./population_1960_2018.csv", as.is=TRUE, header=TRUE)

head(Gen)
head(population)
```

```r
#rename columns to be easier to reference
colnames(Gen)[colnames(Gen)=='X']<- "year"
colnames(Gen)[colnames(Gen)=='Yard.trimmings']<- "Yard"
colnames(Gen)[colnames(Gen)=='All.other']<- "Other"

colnames(population)[colnames(population)=='POPTOTUSA647NWDB']<-"population"

colnames(Gen)

colnames(population)
```

```r
#merge populations to generation and recycling & composting tables
Gen <- merge(x=Gen, y=population, by="year", all.x=TRUE)
head(Gen)
```

```r
#convert thousand tons to tons and divide by population to get tons per person generated
Gen$WastePP <- (1000*Gen$Total)/Gen$population
head(Gen)
```

```r
#run summary to learn more about variables after cleaning up the data frame
summary(Gen)
```

```r
#use pivot_longer to change from each type as a column to each type as an entry per year
#this will make it easier to create a stacked bar chart of total waste split into type by year, include total waste for later step
GenLong <- Gen %>% select(-population, -WastePP) %>%pivot_longer(cols=-c(year, Total), names_to="WasteType", values_to="Amount")
head(GenLong, 15)
```

```r
#calculate the percentage each waste type is of the total per year
GenLong$PercTotal <- (GenLong$Amount/GenLong$Total)*100
head(GenLong)
```

```r
#access the colors in the brewer palette Set3
#I like these colors but want to ensure each waste type is the same color in all graphs
RColorBrewer::brewer.pal(8, "Set3")
```

```r
#plot a stacked bar chart of all waste generated, assign  colors to each type, change axis labels, add labels, and change legend position
png(filename="stackedbar.png")
ggplot(GenLong, aes(x=year, y=Amount, fill=WasteType, color=WasteType))+geom_bar(stat="identity", position="stack")+
scale_fill_manual(values=c("Food"="#8DD3C7", "Glass"="#FCCDE5", "Metals"="#BEBADA", "Other"="#FB8072", "Paper"="#80B1D3", "Plastics"="#FDB462", "Yard"="#B3DE69")) + 
scale_color_manual(values=c("Food"="#8DD3C7", "Glass"="#FCCDE5", "Metals"="#BEBADA", "Other"="#FB8072", "Paper"="#80B1D3", "Plastics"="#FDB462", "Yard"="#B3DE69")) +
labs(title="Waste per Year Broken Down by Type", fill="Type of Waste", x="Year", y="Waste Generated in 1,000 tons") + 
theme_minimal() + guides(color="none") + scale_x_continuous(breaks = c(1960, 1970, 1980, 1990, 2000, 2010, 2020)) +
scale_y_continuous(breaks=c(0, 50000, 100000, 150000, 200000, 250000, 300000))+ 
theme(legend.position = "bottom", legend.justification=c("left", "bottom"))+ theme(text=element_text(size=15))
dev.off()
```

```r
#group by year and use slice_max to find the max waste type per year
#assign less ambiguous column names so that it can be merged with the minimum data frame created next without any confusion
GenMax <- GenLong %>% group_by(year) %>% slice_max(Amount, n=1, with_ties=FALSE)
colnames(GenMax)[colnames(GenMax)=='WasteType']<- 'MaxWT'
colnames(GenMax)[colnames(GenMax)=='Amount']<- 'MaxAmount'
head(GenMax)
```

```r
#group by year and use slice_min to find the minimum waste type per year
#assign less ambiguous column names like above to allow for easier merging in the next step
GenMin <- GenLong %>% group_by(year) %>% slice_min(Amount, n=1, with_ties=FALSE)
colnames(GenMin)[colnames(GenMin)=='WasteType']<- 'MinWT'
colnames(GenMin)[colnames(GenMin)=='Amount']<- 'MinAmount'
head(GenMin)
```

```r
png(filename="percentchange.png")
ggplot(GenLong)+geom_line(aes(x=year, y=PercTotal, color=WasteType)) +theme_minimal() + 
scale_color_manual(values=c("Food"="#8DD3C7", "Glass"="#FCCDE5", "Metals"="#BEBADA", "Other"="#FB8072", "Paper"="#80B1D3", "Plastics"="#FDB462", "Yard"="#B3DE69")) +
labs(title= "Change in Waste Types by Year",x="Year", y="Percent of Total Waste", color="Type of Waste") + theme(text=element_text(size=15))
dev.off()
```

```r
#merge the GenMax and GenMin data frames to be able plot them on the same plot easily
GMinMax<-merge(x=GenMax, y=GenMin, by="year", all.x=TRUE, all.y=TRUE)
head(GMinMax)
```

```r
#plot the minimum and maximum amounts, color corresponds to type and shape corresponds to minimum or maximum amount
png(filename="minmax.png")
ggplot(GMinMax)  + geom_point(aes(x=year, y=MaxAmount, color=MaxWT, shape="Maximum")) +
geom_point(aes(x=year, y=MinAmount, color=MinWT, shape="Minimum"))+theme_minimal()+
labs(title="Maximum & Minimum Waste Type Per Year", 
x="Year", y="Waste in 1,000 tons", color="Type of Waste", shape="") + 
scale_color_manual(values=c("Paper"="#80B1D3","Glass"="#FCCDE5","Plastics"="#FDB462"))+ 
theme(legend.position = "bottom", , legend.box="vertical")+ theme(text=element_text(size=15))
dev.off()
```

# Common Pitfalls / Inefficiencies

**BEFORE you submit your end product, pay attention to the following and check your visualizations and story accordingly.**

**Think of this as a rubric; you will lose points if you don't follow all the principles we have seen in this course.**

---

### - Remember the design analysis questions:

**what, how, who, why?** 

- What are you showing/telling? 
- How are you going to show it (efficiency? visual channels, colors, tasks like comparison, trends, query, etc.) 
- Who is your audience? Why should they care about what you have to tell? What are the **actionable** information they can take from your story? 
- Why are you telling the story? What is your angle, message, point? (make it clear in text AND visuals) 

---

### - Pay attention to the color schemes/palettes:

 - Do NOT use sequential palettes for data attributes that have **no** ordering.
 - Do NOT use qualitative color schemes for attributes where order **matters**. 

---

###  - Color coordinate among plots so that the same data attributes have the same color in all plots:

You can do that by creating a manual palette and use it everywhere to make sure same color is always assigned to the same attribute value (especially important for categories). Do not change color of the same category among plots!

---

###  - Perceptual grouping is important! group and/or order values:

People tend to make sense of things and compare better when they are ordered or grouped:  

- In a bar chart, **order the axes** by time, value, etc. to make comparisons more intuitive. 

- Pie chart: order wedges. 

- In a bubble map, if you want to show trends, use grouping variable to **connect** bubbles, etc. 

---

### - No lengthy text in x axis:

- **Flip** the axes of a bar chart so that character values (text) are **readable on the y axis** left to right, 

**or** 

- at least, put them slanted in x axis for better readibility. 

---

### - Plot only those data columns and/or subsets that tell a clear story:

 - **No crowded** scatter plots with useless linear regression lines that don't tell anything.  
 - You do **NOT** have to plot **ALL** your data (most of the time, you should not). 
 - Look at alternative plots to tell your story effectively. 
 - Your plots should be perceived easily and intuitively. 
  
  

---

### - If you have too many classes (discrete variables/categories), try to GROUP them:

- Instead of giving a color to each of the dozen or so categories, you can put them in meaningful groups and color code to reduce clutter. 

- For sequential variables, you can do the same thing by **binning** numbers into bins (like in a histogram). 

- **Stepped color palettes** may be more useful than smooth sequential palettes (especially for comparison in choropleth maps) for a large range of values in an attribute. 

- You can **group/cluster data into meaningful chunks** and display those as bubble charts, networks, geographical regions, etc. 

- Do **NOT** dump all the data in the plot unless there is a CLEAR pattern/clustering to it that can be **perceived immediately**. 

---

### - ANNOTATE! ANNOTATE! ANNOTATE!

- **Tell** your audience what they are supposed to see. 

- Use **meaningful** titles with a message/statement, no dry titles. 

- Use nice and concise axis labels (make sure they are large enough).

- Use text annotations (look at the **examples in the lab notebook**).

---

### - THEMING

- **Good looking** charts are more **convincing**! Your audience will have more confidence in your work. 

- **Almost NEVER use the default color palette, default theme, or default fonts.**

- Use a subtle color scheme, nice fonts, interesting glyphs, etc. to make your charts look more **sophisticated**. 

---

### - Use space wisely!

- **Balance the plots** to use space effectively.

- Pay attention to **data-ink ratio**.

- **HAVE A FLOW left to right, top to bottom**. 

- If you must have a line plot or bar plot with a lot of empty space, try to put text annotations with relevant information in them.
